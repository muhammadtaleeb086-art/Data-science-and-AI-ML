## Iterators

Iterators are advanced Python concept that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list = [1,2,3,4,5,6]
for i in my_list:
  print(i)
```

```python
type(my_list)
```

```python
print(my_list)
```

```python
## Iterator
iterator = iter(my_list)
print(type(iterator))
```

```python
## iterate through all the element
next(iterator)
```

```python
iterator = iter(my_list)
```

```python

try :
  print(next(iterator))
except StopIteration :
  print("There are no elements in the iterator ")
```

```python
## String Iterator 
my_string = "Hello"
string_iterator = iter(my_string)

print(next(string_iterator))
print(next(string_iterator))
```

## Generators 

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, Which means they generates values on the fly and do not store them in memory.

```python
def square(n):
  for i in range(3):
    yield i**2
  
```

```python
square(3)
```

```python
for i in square(3):
  print(i)
```

```python
a = square(3)
a
```

```python
next(a)
```

```python
def my_generator():
  yield 1
  yield 2
  yield 3
```

```python
gen = my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen :
  print(val)
```

### Practical Example : Reading Large Files

Generators are particularly useful for reading large files because they allow you to process one line at a time without laoding the entire file into memory. 

```python
### Practical : Reading Large files

def read_large_file(file_path):
  with open(file_path , 'r') as file :
    for line in file :
      yield line
```

```python
file_path = 'example.txt'

for line in read_large_file (file_path):
  print(line.strip())
```
