```python
## Question 2 : Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

import numpy as np 

arr1 = np.random.randint(1,21,size=(5,5))
print("Array :")
print(arr1)

border_elements = np.concatenate((arr1[0, :], arr1[-1, :], arr1[1:-1, 0], arr1[1:-1, -1]))
print("Border elements:")
print(border_elements)
```

```python
## Question 2 : Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

import numpy as np

Array = np.arange(1,17).reshape(4,4)
print("Array : ")
print(Array)

row_sum = np.sum(Array,axis=1)
col_sum = np.sum(Array,axis=0)

print("Row Sum :")
print(row_sum)
print("Column Sum :")
print(col_sum)
```

```python
## Question 2 : Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

import numpy as np 

Array1 = np.arange(1,10).reshape((3,3))
print("Array :")
print(Array1)

mean = np.mean(Array1)
std_dev = np.std(Array1)
Normalization = (Array1 - mean ) / std_dev

print("Normalization:")
print(Normalization)
```

```python
## Assignment 5 : 
## Question 1 : Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.

import numpy as np
Arr9 = np.random.randint(1,11,size=(3,3))
row_Arr = np.random.randint(1,11,size=(3,))
print("Original Array :")
print(Arr9)
print("1-D Array :")
print(row_Arr)

result = Arr9 + row_Arr

print("Final Array :")
print(result)


```

```python
## Question 2 : Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

import numpy as np
Arr9 = np.random.randint(1,11,size=(4,4))
col_Arr = np.random.randint(1,11,size=(4,))
print("Original Array :")
print(Arr9)
print("1-D Array :")
print(col_Arr)

result = Arr9 - col_Arr[:, np.newaxis]

print("Final Array :")
print(result)
```
