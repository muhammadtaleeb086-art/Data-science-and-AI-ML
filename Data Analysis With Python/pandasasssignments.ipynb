```python
## Assignment 1 : 
## Question 1 : Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Set the index to be the first column.
import pandas as pd 
import random
import numpy as np

df = pd.DataFrame(np.random.randint(1,100,size=(6,4)),columns=['A','B','C','D'])
print("Original DataFrame")
print(df)

df.set_index('A',inplace=True)
print("Dataframe with new index")
print(df)
```

```python
## Question 2 : Create a Pandas DataFrame with columns 'A', 'B', 'C' and index 'X', 'Y', 'Z'. Fill the DataFrame with random integers and access the element at row 'Y' and column 'B'.

df = pd.DataFrame(np.random.randint(1,100,size=(3,3)),columns=['A','B','C'],index=['X','Y','Z'])
print("Original Dataframe :")
print(df)

element = df.at['Y','B']
print(f"Element at row 'Y' and column 'B' :{element} ")
```

```python
## Assignment 2 :
## Question 1 : Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

df = pd.DataFrame(np.random.randint(1,100,size=(5,3)),columns=['A','B','C'])
print("Original DataFrame :")
print(df)

df['D'] = df['A'] * df['B']
print("Modified Dataframe :")
print(df)
```

```python
## Question 2 : Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

df = pd.DataFrame(np.random.randint(1,100,size=(4,3)),columns=['A','B','C'])
print("Original DataFrame :")
print(df)

row_sum  = df.sum(axis=1)
col_sum = df.sum(axis=0)

print("Row-wise Sum :")
print(row_sum)
print("Column-wise Sum :")
print(col_sum)
```

```python
## Assignment 3 : 
## Question 1 : Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

df = pd.DataFrame(np.random.randint(1,100,size=(5,3)),columns=['A','B','C'])
print("Original Dataframe :")
print(df)

df.iloc[0,1] = np.nan
df.iloc[2,2] = np.nan
df.iloc[4,0] = np.nan

print("Data Frame with NaN:")
print(df)

df.fillna(df.mean(),inplace=True)
print("DataFrame with NaN values filled:")
print(df)
```

```python
## Question 2 : Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.
import pandas as pd 
import numpy as np 
df = pd.DataFrame(np.random.randint(1,100,size=(6,4)),columns=['A','B','C','D'])
print("Original Dataframe :")
print(df)

df.iloc[0,1] = np.nan
df.iloc[2,2] = np.nan
df.iloc[5,2] = np.nan 
print("DataFrame with NaN :")
print(df)

df.dropna(inplace=True)
print("DataFrame with NaN values dropped:")
print(df)
```

```python
## Assignment 4 : 
## Question 1 : Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

df = pd.DataFrame({'Category':np.random.choice(['A','B','C','D'],size=10) , 'Value':np.random.randint(1,100,size=10)})
print("Original Dataframe :")
print(df)

grouped = df.groupby('Category')['Value'].agg(['sum','mean'])
print("Grouped Dataframe :")
print(grouped)
```

```python
# Question 2 : Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.
df = pd.DataFrame({'Product':np.random.choice(['Prod1','Prod2','Prod3'],size=10),'Sales':np.random.choice(['A','B','C'],size=10),'Category':np.random.randint(1,100,size=10)})
print("Original Dataframe :")
print(df)

grouped = df.groupby('Category')['Sales'].sum()
print("Grouped DataFrame :")
print(grouped)
```

```python
## Assignment 5 :
## Question 1 : Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.
import pandas as pd 
import numpy as np

df1 = pd.DataFrame({'Key':['A','B','C','D'],'Value1':np.random.randint(1,100,size=4)})
df2 = pd.DataFrame({'Key':['A','B','C','D'],'Value2':np.random.randint(1,100,size=4)})
print("DataFrame 1 :")
print(df1)
print("DataFrame 2 :")
print(df2)

merged_df = pd.merge(df1,df1,on='Key')
print("Merged DataFrame :")
print(merged_df)
```

```python
## Question 2 : Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.
df1 = pd.DataFrame({'A': np.random.randint(1,100,size=3),'B':np.random.randint(1,100,size=3)})
df2 = pd.DataFrame({'C': np.random.randint(1,100,size=3),'D': np.random.randint(1,100,size=3)})

print("DataFrame 1 : ")
print(df1)
print("DataFrame 2 :")
print(df2)


concat_rows = pd.concat([df1, df2], axis=0)
print("Concatenated DataFrame (rows):")
print(concat_rows)


concat_columns = pd.concat([df1, df2], axis=1)
print("Concatenated DataFrame (columns):")
print(concat_columns)
```

```python
## Assignment 6 :
## Question 1 : Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.
import pandas as pd
import numpy as np 


date_rng = pd.date_range(start='2022-01-01', end='2022-12-31', freq='D')
df = pd.DataFrame(date_rng, columns=['date'])
df['data'] = np.random.randint(0, 100, size=(len(date_rng)))
df.set_index('date', inplace=True)
print("Original DataFrame:")
print(df)

monthly_mean = df.resample('ME').mean()
print("Monthly mean DataFrame:")
print(monthly_mean)
```

```python
# Question 2 : Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.

import pandas as pd 
import numpy as np


date_rng = pd.date_range(start='2021-01-01',end='2021-12-31',freq='D')
df = pd.DataFrame(date_rng,columns=['date'])
df['data'] = np.random.randint(0,100,size=len(date_rng))
df.set_index('date',inplace=True)
print("Original Data Frame :")
print(df)

rolling_mean = df.rolling(window=7).mean()
print("Rolling mean Data Frame :")
print(rolling_mean)
```
