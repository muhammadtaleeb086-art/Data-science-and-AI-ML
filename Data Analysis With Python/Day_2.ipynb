# Pandas-DataFrame And Series

Pandas is a powerful data manipulation library in Python, wisely used for data analysis and data cleaning.It provides two primary data structures:Series and DataFrame. A series is a one-dimensional array-like object,while a Dataframe is a two-dimensional , size-mutable , and potentially heterogeneous tabular data structure with labeled axes (rows and coloumns).


```python
import pandas as pd
```

```python
## Series
## A Pandas Series is a one-deimensional array-like object that can hold any data type. It is similar to a column in a table.

import pandas as pd
data =[1,2,3,4,5]
series = pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
## Create a series fro, Dictionary 
data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)


```

```python
data =[10,20,30]
index = ['a','b','c']
pd.Series(data,index=index)
```

```python
## DataFrame
## create a Dataframe from a Dictionary of list

data = {
  'Name':['Taleeb','Kaifudaa','azizudaa','danisya'],
  'Age':[21,99,5,20],
  'City':['Dubai','Natikybaada','Jopda','Dubai']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
## Create a DataFrame from a list of Dictionaries

data = [
  {'Name':'Taleeb','Age':30,'City':'Saudi'},
  {'Name':'Danish','Age':29,'City':'Dubai'},
  {'Name':'kaif','Age':32,'City':'Newyork'},
  {'Name':'Aziz','Age':31,'City':'Canada'}

]
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Adding a new column to the DataFrame
df['salary'] = [450000,470000,430000,500000] 
df
```

```python
# removing a column from the DataFrame
df.drop('salary',axis=1)
```

```python
df # The original DataFrame remains unchanged after dropping the column, unless we specify inplace=True.
```

```python
df.drop('salary',axis=1,inplace=True) # This will permanently remove the 'salary' column from the DataFrame
df
```
