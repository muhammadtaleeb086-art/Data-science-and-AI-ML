# SQL And SQLite

SQL(Structured Query Language) is a standard language for managing ans manipulating relational databases.SQLite is a self-contained , serverless , and Zero-configuration databases engine that is widely used for embedded database systems. In this lessom , we will cover the basics of SQL and SQLite , including creating databases,tables, and performing various SQL operations.

```python
import sqlite3
```

```python
## Connect to an SQLite database (or create it if it doesn't exist)
connection = sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
## Create a Table
cursor.execute('''
Create Table If Not Exists employees(
     id Integer Primary Key,
     name Text Not Null,
     age Integer,
     department text
     )
''')

## Commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

```python
## Insert the data in sqlite table
cursor.execute('''
Insert Into employees(name,age,deparment)
               values('Taleeb',32,'Data Scientist')
''')

cursor.execute('''
Insert Into employees(name,age,deparment)
               values('kaif',34,'AI engineer')
''')


cursor.execute('''
Insert Into employees(name,age,deparment)
               values('Aziz',30,'Al')
''')

## Commit the changes
connection.commit()
```

```python
## Query the Data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## Print the Queried Data

for row in rows:
  print(row)
```

```python
## Update the data in the table 
cursor.execute('''
UPDATE employees
               Set age = 34
               where name = "Taleeb"
''')

connection.commit()
```

```python
## Query the Data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## Print the Queried Data

for row in rows:
  print(row)
```

```python
## Delete the data from the table
cursor.execute('''
Delete from employees
               Where name = 'Aziz'
''')
```

```python
## Query the Data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## Print the Queried Data

for row in rows:
  print(row)
```

```python
## Working with Sales Data 
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()


## Create a table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
   id INTEGER PRIMARY KEY,
   date TEXT NOT NULL,
   product TEXT NOT NULL,
   sales INTEGER,
   region TEXT
   )

''')

## INSERT DATA INTO THE SALES TABLE
sales_data = [
  ('2023-01-01','Product1',100,'North'),
  ('2023-01-02','Product2',200,'South'),
  ('2023-01-03','Product1',150,'East'),
  ('2023-01-04','Product3',250,'West'),
  ('2023-01-05','Product2',150,'North')
]

cursor.executemany('''
INSERT INTO sales(date,product,sales,region)
VALUES(?,?,?,?)
''',sales_data)

## Commit the changes
connection.commit()
```

```python
## Query the Data from the table
cursor.execute('Select * from sales')
rows=cursor.fetchall()

## Print the Queried Data

for row in rows:
  print(row)
```

```python
## Close the connection
connection.close()
```
